```python
import pandas as pandas
import matplotlib.pyplot as plt
import seaborn as sns

df=pandas.read_csv(r"C:\Users\user\Downloads\archive (4)\Iris.csv")
print("First 5 rows of the database:")
print(df.head())

print("\nLast 5 rows of the database:")
print(df.tail())

print("\nShape of the database:")
print(df.shape)

print("\nColumns of the database:")
print(df.columns)

print(df.isnull().sum())

df=df.dropna()

print(df.isnull().sum())

print(df.describe())

sns.set_style("whitegrid")

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="SepalLengthCm", bins=10)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal length (cm)")
plt.ylabel("Number of Flowers")
plt.show()
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Species",y="SepalLengthCm")
plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.show()
```
